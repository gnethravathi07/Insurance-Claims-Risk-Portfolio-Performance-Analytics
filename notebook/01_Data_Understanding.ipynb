```python
import pandas as pd
import numpy as np
from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 150)

print("Pandas Version:", pd.__version__)
print("NumPy Version:", np.__version__)
print("Libraries imported successfully!")
```

```python
from pathlib import Path

data_path = Path("../data")

customers = pd.read_csv(data_path / "customers.csv")
policies = pd.read_csv(data_path / "policies.csv")
claims = pd.read_csv(data_path / "claims.csv")
payments = pd.read_csv(data_path / "payments.csv")
risk_assessment = pd.read_csv(data_path / "risk_assessment.csv")
agents = pd.read_csv(data_path / "agents.csv")
policy_agents = pd.read_csv(data_path / "policy_agents.csv")
vehicles = pd.read_csv(data_path / "vehicles.csv")

print("All 8 datasets loaded successfully!")
```

```python
### Dataset Dimensions

datasets = {
    "Customers": customers,
    "Policies": policies,
    "Claims": claims,
    "Payments": payments,
    "Risk Assessment": risk_assessment,
    "Agents": agents,
    "Policy Agents": policy_agents,
    "Vehicles": vehicles
}

dataset_summary = pd.DataFrame([
    {
        "Dataset": name,
        "Rows": df.shape[0],
        "Columns": df.shape[1]
    }
    for name, df in datasets.items()
])

display(dataset_summary)
```

```python
for name, df in datasets.items():
    print(f"\n{'=' * 50}")
    print(f"Dataset: {name}")
    print(f"{'=' * 50}")

    df.info()
```

```python
column_summary = pd.DataFrame([
    {
        "Dataset": name,
        "Column": col,
        "Data_Type": str(df[col].dtype),
        "Non_Null_Count": df[col].count()
    }
    for name, df in datasets.items()
    for col in df.columns
])

display(column_summary)
```

```python
for name, df in datasets.items():
    print(f"\nDataset: {name}")
    display(df.head(5))
```

```python
## Examine Categorical Values

categorical_columns = {
    "Customers": ["Gender", "Region"],
    "Policies": ["Policy_Type", "Policy_Status"],
    "Claims": ["Claim_Status", "Fraud_Flag", "Claim_Type"],
    "Payments": ["Payment_Method"],
    "Risk Assessment": ["Risk_Category"],
    "Agents": ["Region", "Channel"],
    "Vehicles": ["Vehicle_Type"]
}

for name, columns in categorical_columns.items():
    print(f"\n{'=' * 50}")
    print(f"Dataset: {name}")
    print(f"{'=' * 50}")

    for col in columns:
        print(f"\n{col}:")
        print(datasets[name][col].value_counts(dropna=False))
```

```python
category_summary = pd.concat(
    [
        df[col].value_counts(dropna=False)
        .rename_axis("Category")
        .reset_index(name="Count")
        .assign(Dataset=name, Column=col)
        for name, columns in categorical_columns.items()
        for col in columns
        for df in [datasets[name]]
    ],
    ignore_index=True
)

category_summary = category_summary[
    ["Dataset", "Column", "Category", "Count"]
]

display(category_summary)
```

```python
numerical_columns = {
    "Customers": ["Age", "Annual_Income"],
    "Policies": ["Annual_Premium", "Coverage_Amount"],
    "Claims": ["Claim_Amount"],
    "Payments": ["Payment_Amount"],
    "Risk Assessment": ["Risk_Score"],
    "Agents": ["Experience_Years"],
    "Vehicles": ["Vehicle_Age", "Insured_Value"]
}

numerical_summary = pd.concat(
    [
        df[columns].describe().T.assign(Dataset=name)
        for name, columns in numerical_columns.items()
        for df in [datasets[name]]
    ]
)

numerical_summary = numerical_summary.reset_index(
    names="Column"
)

numerical_summary = numerical_summary[
    [
        "Dataset", "Column", "count",
        "mean", "std", "min",
        "25%", "50%", "75%", "max"
    ]
]

display(numerical_summary.round(2))
```

```python
date_columns = {
    "Customers": ["Join_Date"],
    "Policies": ["Start_Date", "End_Date"],
    "Claims": ["Claim_Date"],
    "Payments": ["Payment_Date"],
    "Risk Assessment": ["Assessment_Date"]
}

date_summary = []

for name, columns in date_columns.items():
    for col in columns:
        dates = pd.to_datetime(
            datasets[name][col],
            errors="coerce"
        )

        date_summary.append({
            "Dataset": name,
            "Date_Column": col,
            "Earliest_Date": dates.min(),
            "Latest_Date": dates.max(),
            "Invalid_Dates": dates.isna().sum()
        })

date_summary = pd.DataFrame(date_summary)

display(date_summary)
```

```python
# Check Key Uniqueness

key_columns = {
    "Customers": ["Customer_ID"],
    "Policies": ["Policy_ID"],
    "Claims": ["Claim_ID"],
    "Payments": ["Payment_ID"],
    "Risk Assessment": ["Customer_ID"],
    "Agents": ["Agent_ID"],
    "Policy Agents": ["Policy_ID"],
    "Vehicles": ["Vehicle_ID"]
}

key_summary = pd.DataFrame([
    {
        "Dataset": name,
        "Total_Rows": len(datasets[name]),
        "Unique_Keys": len(
            datasets[name][columns].drop_duplicates()
        ),
        "Duplicate_Keys": datasets[name].duplicated(
            subset=columns
        ).sum()
    }
    for name, columns in key_columns.items()
])

display(key_summary)
```

```python
# Validate Foreign Key Relationships

relationships = [
    ("Policies", "Customer_ID", "Customers", "Customer_ID"),
    ("Claims", "Policy_ID", "Policies", "Policy_ID"),
    ("Payments", "Claim_ID", "Claims", "Claim_ID"),
    ("Risk Assessment", "Customer_ID", "Customers", "Customer_ID"),
    ("Policy Agents", "Policy_ID", "Policies", "Policy_ID"),
    ("Policy Agents", "Agent_ID", "Agents", "Agent_ID"),
    ("Vehicles", "Customer_ID", "Customers", "Customer_ID")
]

fk_results = []

for child, child_key, parent, parent_key in relationships:

    invalid_count = (
        ~datasets[child][child_key].isin(
            datasets[parent][parent_key]
        )
    ).sum()

    fk_results.append({
        "Child_Table": child,
        "Foreign_Key": child_key,
        "Parent_Table": parent,
        "Invalid_References": invalid_count,
        "Status": "PASS" if invalid_count == 0 else "FAIL"
    })

fk_summary = pd.DataFrame(fk_results)

display(fk_summary)
```

```python
# Examine Dataset Relationship Cardinality

relationship_summary = pd.DataFrame([
    {
        "Relationship": "Customers → Policies",
        "Parent_Records": customers["Customer_ID"].nunique(),
        "Child_Records": len(policies),
        "Unique_Parents_Referenced": policies["Customer_ID"].nunique()
    },
    {
        "Relationship": "Policies → Claims",
        "Parent_Records": policies["Policy_ID"].nunique(),
        "Child_Records": len(claims),
        "Unique_Parents_Referenced": claims["Policy_ID"].nunique()
    },
    {
        "Relationship": "Claims → Payments",
        "Parent_Records": claims["Claim_ID"].nunique(),
        "Child_Records": len(payments),
        "Unique_Parents_Referenced": payments["Claim_ID"].nunique()
    }
])

relationship_summary["Parents_Without_Children"] = (
    relationship_summary["Parent_Records"]
    - relationship_summary["Unique_Parents_Referenced"]
)

display(relationship_summary)
```

```python
# Final Data Understanding Summary

final_summary = pd.DataFrame([
    {
        "Dataset": name,
        "Total_Rows": len(df),
        "Total_Columns": df.shape[1],
        "Missing_Values": df.isna().sum().sum(),
        "Duplicate_Rows": df.duplicated().sum()
    }
    for name, df in datasets.items()
])

display(final_summary)

print("\nTotal Records:", final_summary["Total_Rows"].sum())
print("Total Missing Values:", final_summary["Missing_Values"].sum())
print("Total Duplicate Rows:", final_summary["Duplicate_Rows"].sum())
```


### Key Findings

- Validated 7 foreign key relationships with no invalid references.
